```python
import os
import joblib
from google.colab import drive
from sklearn.metrics import accuracy_score
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier

# ==============================================================================
# 1. LIGAR AO GOOGLE DRIVE [M]
# ==============================================================================
drive.mount('/content/drive')
drive_path = '/content/drive/MyDrive/Projeto_Cybersecurity'

# Garante que a pasta 'models' existe no Drive
os.makedirs(f'{drive_path}/models', exist_ok=True)

# ==============================================================================
# 2. CARREGAR OS DADOS PROCESSADOS (Gerados pelo Data Preparation) [M]
# ==============================================================================
print("A carregar dados pré-processados do Google Drive...")
X_train_scaled = joblib.load(f'{drive_path}/models/X_train_scaled.pkl')
X_test_scaled = joblib.load(f'{drive_path}/models/X_test_scaled.pkl')
y_train = joblib.load(f'{drive_path}/models/y_train.pkl')
y_test = joblib.load(f'{drive_path}/models/y_test.pkl')

print("✅ Dados de Treino e Teste carregados com sucesso!")

# ==============================================================================
# 3. TREINAR E COMPARAR MÚLTIPLOS MODELOS (Supervised Learning) [M]
# ==============================================================================
# Expandido para 4 modelos para satisfazer a diversidade exigida no programa
models = {
    'Decision Tree': DecisionTreeClassifier(random_state=42),
    'Random Forest': RandomForestClassifier(n_estimators=100, random_state=42),
    'Logistic Regression': LogisticRegression(max_iter=1000, random_state=42),
    'KNN': KNeighborsClassifier(n_neighbors=5)
}

print("\n==========================================")
print("=== INÍCIO DO TREINO DOS MODELOS ===")
print("==========================================")

for name, model in models.items():
    print(f"\nA treinar {name}...")
    model.fit(X_train_scaled, y_train)

    # Avaliação rápida de Accuracy
    y_pred = model.predict(X_test_scaled)
    acc = accuracy_score(y_test, y_pred)
    print(f"Accuracy de {name}: {acc:.4f}")

    # Guardar o modelo treinado no Google Drive [M]
    filename = f"{drive_path}/models/{name.lower().replace(' ', '_')}_model.pkl"
    joblib.dump(model, filename)
    print(f"✅ Modelo guardado em: '{filename}'")

print("\n==========================================")
print("✅ MODELAÇÃO CONCLUÍDA COM SUCESSO!")
print("Todos os 4 modelos foram salvos na pasta 'models/' do Google Drive.")
print("==========================================")
```
